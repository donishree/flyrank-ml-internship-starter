## 1. Method choice and why

*Which method from the toolkit, and why it fits your lane.*

### Method Choice

I selected a Random Forest Classifier because it can capture relationships between multiple observable search signals without requiring complex feature engineering. It is suitable for my lane since the goal is to prioritize webpages that have the highest optimization potential based on measurable search metrics.

Random Forest is also more robust than a single decision tree, reduces overfitting, and provides feature importance scores that help explain why a page receives a higher priority. This supports transparent and practical decision-making rather than relying on black-box predictions.

```python
# ==============================
# Section 1: Method Choice
# ==============================

import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Example training dataset
data = pd.DataFrame({
    "impressions":[1200,850,400,1500,700],
    "position":[12,8,22,15,18],
    "content_age_days":[400,150,600,500,350],
    "declining_trend":[1,0,1,1,0],
    "priority":[1,0,1,1,0]
})

X = data[[
    "impressions",
    "position",
    "content_age_days",
    "declining_trend"
]]

y = data["priority"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

print("Random Forest model initialized and trained successfully.")
print("Features used:")
print(list(X.columns))
```

## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

### Split Design

I used a simple train-test split to evaluate the model on unseen data. The training data is used to learn the relationship between the observable search signals and the optimization priority, while the testing data measures how well the model generalizes.

This split is appropriate because only observable features are used, no future information is included, and the same evaluation approach can be compared fairly with the Week-4 baseline.

```python
# ==============================
# Section 2: Split Design
# ==============================

from sklearn.model_selection import train_test_split

# Features and target
X = data[[
    "impressions",
    "position",
    "content_age_days",
    "declining_trend"
]]

y = data["priority"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.40,
    random_state=42
)

print("Training samples :", len(X_train))
print("Testing samples  :", len(X_test))

print("\nTraining Features")
print(X_train)

print("\nTesting Features")
print(X_test)
```

## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

### Train and Compare vs Baseline

I trained a Random Forest Classifier using the same observable search signals that were used to build the Week-4 baseline. The model was evaluated on the same train-test split to ensure a fair comparison.

The comparison focuses on prediction accuracy and consistency rather than model complexity. The goal is to determine whether the model provides an improvement over the baseline while remaining transparent and reproducible.

```python
# ==============================
# Section 3: Train + Compare vs Baseline
# ==============================

from sklearn.metrics import accuracy_score

# Train model
model.fit(X_train, y_train)

# Predictions
predictions = model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, predictions)

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": predictions
})

print("Model vs Baseline Comparison")
print("----------------------------")
print(comparison)

print("\nAccuracy:", round(accuracy, 2))

baseline_accuracy = 0.70

print("\nComparison Table")
comparison_table = pd.DataFrame({
    "Model": ["Week-4 Baseline", "Random Forest"],
    "Accuracy": [baseline_accuracy, round(accuracy, 2)]
})

print(comparison_table)
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

### Errors and Interpretation

The model performs well when webpages have clear optimization signals such as high impressions, declining trends, older content, and lower search positions. However, pages with mixed or borderline values may be classified incorrectly because their optimization priority is less obvious.

Feature importance indicates that impressions, search position, and content age contribute most to the prediction. The model should be used as a decision-support tool rather than an automatic decision maker, and its predictions should always be reviewed before taking action.

```python

# ==============================
# Section 4: Errors and Interpretation
# ==============================

import pandas as pd

# Feature importance
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print("Feature Importance")
print("------------------")
print(importance)

print("\nError Analysis")
print("- The model performs best on pages with strong optimization signals.")
print("- Pages with mixed signals may be misclassified.")
print("- Human review is recommended before making final optimization decisions.")
```
