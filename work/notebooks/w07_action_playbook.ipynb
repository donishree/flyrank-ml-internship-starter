## Ranked Actions + Reason Codes

The playbook ranks content optimization actions using the baseline model scores and observable search signals. Each recommendation includes a clear reason code so that reviewers understand why the page was selected.

### Ranked Actions

1. Refresh outdated content with high impressions.
2. Improve pages showing declining performance.
3. Optimize pages with poor average search position.
4. Review pages where multiple signals indicate optimization opportunity.
5. Monitor lower-priority pages before making changes.

### Reason Codes

- HIGH_IMPRESSIONS – The page receives enough visibility to justify optimization.
- STALE_CONTENT – The content has not been refreshed recently.
- DECLINING_TREND – Performance has decreased over time.
- LOW_POSITION – The page ranks lower than desired.
- REVIEW_PRIORITY – Multiple signals suggest the page should be reviewed first.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human Review and No-Go List

All recommendations generated by the playbook require human validation before implementation. The model is intended to support decision-making rather than automate content updates.

### Human Review Rules
- Verify that the recommendation matches the current content.
- Confirm that optimization aligns with business and editorial goals.
- Review search performance using recent data before taking action.
- Document the final decision for each recommendation.

### No-Go Cases
- Do not automatically publish AI-generated content.
- Do not modify high-performing pages without evidence.
- Do not remove valuable content based only on model scores.
- Do not use confidential or private information in recommendations.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring and Retrain Triggers

The playbook should be monitored regularly to ensure recommendations remain accurate and relevant. Retraining or reviewing the model should be considered when performance or search behavior changes significantly.

### Monitoring Triggers
- Monitor changes in impressions and average position.
- Track click-through rate (CTR) trends.
- Observe changes in content freshness.
- Review model recommendations periodically.

### Retrain Triggers
- Significant drop in prediction performance.
- Major changes in search engine behavior.
- Large increases in new content or pages.
- Noticeable data drift in search signals.

Regular monitoring helps maintain reliable recommendations while keeping the playbook aligned with current search performance.

```python
import pandas as pd

monitoring = pd.DataFrame({
    "Trigger": [
        "Drop in model performance",
        "Change in search trends",
        "Content growth",
        "Data drift detected",
        "Scheduled monthly review"
    ],
    "Action": [
        "Retrain model",
        "Review recommendations",
        "Recalculate scores",
        "Retrain and validate",
        "Monitor performance"
    ]
})

print("Monitoring and Retrain Triggers")
display(monitoring)
```
